```python
!echo "Day la du lieu can duoc bao mat" > plaintext.txt

!openssl enc -aes-256-cbc -salt -pbkdf2 -in plaintext.txt -out encrypted.bin -k "mat_khau_bi_mat_123"

!echo "--- DỮ LIỆU ĐÃ MÃ HÓA (BASE64) ---"
!base64 encrypted.bin
!echo ""

!openssl enc -d -aes-256-cbc -salt -pbkdf2 -in encrypted.bin -out decrypted.txt -k "mat_khau_bi_mat_123"

!echo "--- DỮ LIỆU SAU KHI GIẢI MÃ ---"
!cat decrypted.txt
```
