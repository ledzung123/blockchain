```python
!openssl genrsa -out private_key.pem 2048

!openssl rsa -in private_key.pem -pubout -out public_key.pem

!echo "Thong tin sinh vien truong dai hoc cong nghe" > rsa_plaintext.txt

!openssl pkeyutl -encrypt -pubin -inkey public_key.pem -in rsa_plaintext.txt -out rsa_encrypted.bin

!echo "--- DỮ LIỆU ĐÃ MÃ HÓA BẰNG RSA PUBLIC KEY (BASE64) ---"
!base64 rsa_encrypted.bin
!echo ""

!openssl pkeyutl -decrypt -inkey private_key.pem -in rsa_encrypted.bin -out rsa_decrypted.txt

!echo "--- DỮ LIỆU SAU KHI GIẢI MÃ BẰNG RSA PRIVATE KEY ---"
!cat rsa_decrypted.txt
```
