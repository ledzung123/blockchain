```python
# 2. Mã hoá và giải mã bằng public key RSA
!openssl genrsa -out private_key.pem 2048

!openssl rsa -in private_key.pem -pubout -out public_key.pem

!echo "Thong tin sinh vien truong dai hoc cong nghe" > rsa_plaintext.txt

!openssl pkeyutl -encrypt -pubin -inkey public_key.pem -in rsa_plaintext.txt -out rsa_encrypted.bin

!echo "Dữ liệu đã mã hoá"
!base64 rsa_encrypted.bin
!echo ""

!openssl pkeyutl -decrypt -inkey private_key.pem -in rsa_encrypted.bin -out rsa_decrypted.txt

!echo "Dữ liệu sau giải mã"
!cat rsa_decrypted.txt
```

```python
# 1. Mã hoá và giải mã sử dụng AES 256

!echo "Day la du lieu can duoc bao mat" > plaintext.txt

!openssl enc -aes-256-cbc -salt -pbkdf2 -in plaintext.txt -out encrypted.bin -k "mat_khau_bi_mat_123"

!echo "Dữ liệu đã mã hoá"
!base64 encrypted.bin
!echo ""

!openssl enc -d -aes-256-cbc -salt -pbkdf2 -in encrypted.bin -out decrypted.txt -k "mat_khau_bi_mat_123"

!echo "Dữ liệu sau giải mã"
!cat decrypted.txt
```
