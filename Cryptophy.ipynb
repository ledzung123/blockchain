```python
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import padding
from cryptography.hazmat.primitives import serialization

with open("public_key.pem", "rb") as key_file:
    public_key = serialization.load_pem_public_key(key_file.read())

with open("course_cancel.txt", "rb") as f:
    message = f.read()

with open("signature.bin", "rb") as f:
    signature = f.read()

try:
    public_key.verify(
        signature,
        message,
        padding.PKCS1v15(),
        hashes.SHA256()
    )
    print("KẾT QUẢ: Dữ liệu toàn vẹn, chữ ký hợp lệ")
except Exception as e:
    print("KẾT QUẢ: Dữ liệu bị sửa đổi hoặc chữ ký sai")
```

```python
from cryptography.hazmat.primitives.asymmetric import padding as rsa_padding

with open("private_key.pem", "rb") as key_file:
    private_key = serialization.load_pem_private_key(
        key_file.read(),
        password=None
    )

with open("rsa_encrypted.bin", "rb") as f:
    ciphertext = f.read()

try:
    plaintext = private_key.decrypt(
        ciphertext,
        rsa_padding.PKCS1v15()
    )
    print("Dữ liệu sau giải mã")
    print(plaintext.decode('utf-8'))
except Exception as e:
    print("Giải mã RSA thất bại:", str(e))
```

```python
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.primitives.kdf.pbkdf2 import PBKDF2HMAC
from cryptography.hazmat.primitives import hashes, padding as sym_padding
from cryptography.hazmat.backends import default_backend

with open("encrypted.bin", "rb") as f:
    encrypted_data = f.read()

salt = encrypted_data[8:16]
ciphertext = encrypted_data[16:]
password = b"mat_khau_bi_mat_123"

kdf = PBKDF2HMAC(
    algorithm=hashes.SHA256(),
    length=48,
    salt=salt,
    iterations=10000,
    backend=default_backend()
)
key_iv = kdf.derive(password)
key = key_iv[:32]
iv = key_iv[32:]

cipher = Cipher(algorithms.AES(key), modes.CBC(iv), backend=default_backend())
decryptor = cipher.decryptor()
padded_plaintext = decryptor.update(ciphertext) + decryptor.finalize()


unpadder = sym_padding.PKCS7(algorithms.AES.block_size).unpadder()
plaintext = unpadder.update(padded_plaintext) + unpadder.finalize()

print("Dữ liệu sau giải mã")
print(plaintext.decode('utf-8'))
```
